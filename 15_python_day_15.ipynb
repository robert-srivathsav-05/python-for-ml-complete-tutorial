## 1. Solid Square / Rectangle

**Logic:** A square/rectangle prints the same number of stars in every row. This means both the outer loop (rows) and inner loop (columns) can run a fixed number of times — no relationship between them is needed.

- Outer loop (`row`) → controls how many rows are printed
- Inner loop (`column`) → prints a fixed number of stars in each row
- `end=""` keeps stars on the same line; `print()` after the inner loop moves to the next row


````python
rows = 4
cols = 7

for row in range(rows):
    for column in range(cols):
        print("* ", end="")
    print()
````

## 2. Right-Angled Triangle (Left-Aligned)

**Logic:** Unlike the square, the number of stars now depends on which row you're on. Row 1 has 1 star, row 2 has 2 stars, and so on.

- The inner loop's range is tied to the outer loop variable `i`, using `range(i)`
- This is the key shift from a fixed-size pattern to a **growing** pattern


````python
n = 5

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()
````

## 3. Right-Angled Triangle (Right-Aligned)

**Logic:** This triangle needs leading spaces before the stars, so that every row's stars end at the same right edge.

**Why the `2 ×` in the spacing formula?**
Each star is printed as `"* "` (a star plus a trailing space) — 2 characters wide, not 1. So when row `i` is "missing" `(n - i)` stars compared to the widest row, it's actually missing `2 × (n - i)` characters of width. The leading spaces need to make up for that full width, which is why the formula is `2 * (n - i)` and not just `(n - i)`.


````python
n = 5

for i in range(1, n + 1):
    print(" " * 2 * (n - i), end="")
    for j in range(i):
        print("*", end=" ")
    print()
````

## 4. Inverted Right-Angled Triangle

**Logic:** This is pattern #2, but counting *downward* instead of upward. Row 1 (the widest) has `n` stars, and each following row has one fewer.

- `range(n, 0, -1)` counts down from `n` to `1`
- The inner loop naturally shrinks each row since it depends on the shrinking `i`


````python
n = 5

for i in range(n, 0, -1):
    for j in range(i):
        print("* ", end="")
    print()
````

## 5. Pyramid (Centered Triangle)

**Logic:** This combines leading spaces (like pattern #3) with growing stars (like pattern #2), but the goal is **centering**, not right-alignment.

**Why `(n - i)` spaces work here (no `2 ×` needed):**
Right-alignment requires every row's *right edge* to land at the exact same column — a strict requirement that needed the `2 ×` correction in pattern #3. Centering is a different, looser requirement: each row's star cluster just needs to be symmetric around the same central column as the row above it. When you work through the actual character positions, using plain `(n - i)` spaces already produces stars that are symmetrically centered — the trailing space after each star balances out on the right side automatically. This is a good example of why it's important to check *what* a pattern actually requires (strict edge alignment vs. symmetry) rather than assuming the same formula always applies.


````python
n = 5

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(i):
        print("*", end=" ")
    print()
````

## 6. Inverted Pyramid

**Logic:** This is pattern #5, flipped upside down. The outer loop counts down from `n` to `1`, so the row with the most stars is printed first, and each following row shrinks while the leading spaces grow.


````python
n = 5

for i in range(n, 0, -1):
    print(" " * (n - i), end="")
    for j in range(i):
        print("*", end=" ")
    print()
````

## 7. Diamond

**Logic:** A diamond is simply the pyramid (#5) stacked directly on top of the inverted pyramid (#6).

- The first loop prints rows `1` to `n` (the pyramid half, including the widest row once)
- The second loop prints rows `n - 1` down to `1` — **starting at `n - 1`, not `n`**, so the widest row isn't printed twice


````python
n = 5

# Top half: pyramid
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(i):
        print("*", end=" ")
    print()

# Bottom half: inverted pyramid (starts at n-1 to avoid repeating the widest row)
for i in range(n - 1, 0, -1):
    print(" " * (n - i), end="")
    for j in range(i):
        print("*", end=" ")
    print()
````

## 8. Number Triangle

**Logic:** Structurally identical to pattern #2 (left-aligned triangle) — the only difference is that instead of printing a fixed `"*"`, we print the inner loop variable `j` itself, which naturally counts `1, 2, 3, ...` within each row.


````python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

## 9. Floyd's Triangle

**Logic:** Unlike the number triangle, the numbers here keep increasing **continuously across the whole pattern** instead of resetting to 1 at the start of every row.

- A `value` variable is declared **outside both loops**, so it persists across rows instead of resetting
- Every time a number is printed, `value` is incremented by 1
- The outer loop still controls how many numbers appear per row (`i` numbers in row `i`), it just doesn't control *what* those numbers are


````python
n = 5
value = 1

for i in range(1, n + 1):
    for j in range(i):
        print(value, end=" ")
        value += 1
    print()
````

## 10. Pascal's Triangle

**Logic:** Every number in Pascal's triangle is the sum of the two numbers directly above it (imagining invisible zeroes off both edges of each row). The first and last number in every row is always `1`.

```
Row 1:        1
Row 2:       1 1
Row 3:      1 2 1
Row 4:     1 3 3 1
Row 5:    1 4 6 4 1
```

**Approach used (no factorials required):** each row is built from the **previous row's values**, stored in a list.

- `prev_row` holds the previous row's numbers (starts empty, since row 1 has nothing above it)
- For each row, a new list `curr_row` is built
- If a position is the first or last in the row (`j == 0` or `j == i`), it's always `1`
- Otherwise, it's `prev_row[j-1] + prev_row[j]` — the sum of the two numbers above it, directly matching the rule
- After printing, `prev_row` is updated to `curr_row` so the next row can build on it

This uses a **list** (a datatype you already know) in a slightly more advanced way — building it up dynamically inside a loop rather than declaring it with fixed values upfront. This same idea — using the previous step's result to compute the next step — is a pattern that shows up constantly in ML (e.g. iterative updates, sequence processing), which is why it's worth understanding even though Pascal's triangle itself isn't something you'll use directly in research work.


````python
n = 5
prev_row = []

for i in range(n):
    curr_row = []
    for j in range(i + 1):
        if j == 0 or j == i:
            curr_row.append(1)
        else:
            curr_row.append(prev_row[j - 1] + prev_row[j])
    for num in curr_row:
        print(num, end=" ")
    print()
    prev_row = curr_row
````

## 11. Hollow Square

**Logic:** A full square has stars everywhere. A hollow square only has stars on the **border** — the first row, the last row, and the first/last column of every row in between.

- `i == 0 or i == n - 1` → checks if this is the first or last row
- `j == 0 or j == n - 1` → checks if this is the first or last column
- If either condition is true → print a star; otherwise → print a space


````python
n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

## 12. Hollow Triangle

**Logic:** Similar idea to the hollow square, but trickier because the triangle's shape changes every row instead of staying fixed.

For each row `i` (with `i + 1` positions, same structure as the solid left-triangle), a star is printed only if:
- `j == 0` → it's the first position in the row, **or**
- `j == i` → it's the last position in the row, **or**
- `i == n - 1` → this is the last row (the triangle's base), which is fully solid

Everywhere else → a space.


````python
n = 5

for i in range(n):
    for j in range(i + 1):
        if j == 0 or j == i or i == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

## 13. Alphabet Pattern

**Logic:** Structurally identical to the number triangle (#8) — the only difference is printing letters instead of numbers.

**Approach used:** a list of letters `["A", "B", "C", "D", "E"]` is indexed with the inner loop variable `j`, so row `i` prints letters `alphabet[0]` through `alphabet[i]`.

*(Alternative approach: Python's `chr()` function converts a number into its corresponding character using ASCII codes — e.g. `chr(65)` gives `'A'`. So `chr(65 + j)` would also produce `A, B, C, ...` without needing a list.)*


````python
alphabet = ["A", "B", "C", "D", "E"]

for i in range(len(alphabet)):
    for j in range(i + 1):
        print(alphabet[j], end=" ")
    print()
````
