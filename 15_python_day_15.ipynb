# <u> Nested loops</u>

A nested loop is simply a loop inside another. It runs everytime from start to end everytime the outer loop runs.
That means, if we have 10 classes to take attendance from every class, we dont need to go through every class and then take attendance. We can go to a class, and take attendance of them completely and then go to next one and then we go attendance and like this at a time, we can take 10 classess attendance easily.

For sample nested loop problem, look at the following code:

```python
for i in range(3):
  for j in range(2):
    print(i,j)
```
In the above code, thats what how a nested loop looks like..
The inner loop nested in the outer one, and if we look at the code, the outer loop range is upto 3 which means it runs three times and the inner loop range is upto 2 which means it runs two times. So now how the code works is, first the outer loop range gives starts, gives the element 0, which triggers the run of the inner loop and then the inner loop starts running for two steps because it has range 2, then here in the inner loop, it starts with 0, then here when j=0, i=0, and in the loop the indented code is to print the i,j values, it will print
0,0
and then the j value changes to 1, and again the indented code runs which prints
0,1
And the inner loop ends, and that means, the indented code of the outer block is ended which means the i value changes now from 0 to 1 and again the same continues upto i=2, which makes the outer loop end and which means the program is completed.

````python
for i in range (3):
  for j in range (2):
    print(i,j)
````

The core idea is for every single value in the outer loop, the inner loop runs from start to end and when it ends, its reset and ready to run from start for the next one.

## <u>Nested while loops</u>

The idea is exactly the same as nested for loops — a loop inside a loop — but now both loops need their own counter variable that you manually update, since while doesn't have a built-in counter like range() gives you.

Here's the basic structure, mirroring our very first example:




````python
i = 0
while i < 3:
    j = 0
    while j < 2:
        print(i, j)
        j += 1
    i += 1
````

Notice a few important things:

j = 0 is written inside the outer loop, but outside/above the inner loop. This is critical — it means j gets reset to 0 every time the outer loop runs again. If you wrote j = 0 outside the outer loop (only once, at the very top), j would keep climbing forever across outer iterations and the inner loop would never run again after the first round. j += 1 is inside the inner loop, so j climbs 1 → 2 → exits inner loop. i += 1 is inside the outer loop, but outside the inner loop, so i only climbs once per full inner loop completion.

This output will match your very first nested for loop example exactly:
```
0 0
0 1
1 0
1 1
2 0
2 1
```
The single most common mistake with nested while loops is forgetting to reset the inner counter. Keep that in your head as you solve — if your inner loop only runs once total instead of resetting each time, that's almost always the bug.
